```python
import pandas as pd
import numpy as np

df = pd.read_csv("slm_dataset.csv.csv")

print("Dataset loaded:", df.shape)
```

```python
selected_columns = [
    "program_id",
    "cip_title",
    "cip_family_title",
    "credential_name",
    "median_earnings_4yr_usd",
    "median_earnings_1yr_usd",
    "median_earnings_5yr_usd",
    "median_debt_usd",
    "earnings_trajectory_category",
    "largest_linked_occupation",
    "occupation_typical_entry_education",
    "occupation_growth_pct_2024_34",
    "occupation_annual_openings_thousands",
    "occupation_median_wage_2024_usd",
    "ai_software_occupation_share",
    "expert_system_occupation_share",
    "ai_tool_examples"
]

slm_df = df[selected_columns].copy()

print("Selected records:", slm_df.shape)

display(slm_df.head())
```

```python
slm_df = slm_df.dropna(subset=["cip_title"])

slm_df = slm_df.reset_index(drop=True)

print("Records after cleaning:", len(slm_df))
```

```python
qa_data = []

def add_qa(record_id, program, question, answer):

    qa_data.append({
        "record_id": record_id,
        "program": program,
        "question": question,
        "answer": answer
    })


for idx, row in slm_df.iterrows():

    program = str(row["cip_title"])

    context = (
        f"Educational program: {program}. "
        f"Credential: {row['credential_name']}."
    )

    # Graduate earnings
    if pd.notna(row["median_earnings_4yr_usd"]):
        add_qa(
            idx,
            program,
            f"{context} What are the median graduate earnings after four years?",
            f"The reported median graduate earnings after four years are "
            f"${row['median_earnings_4yr_usd']:,.0f} USD."
        )

    # Student debt
    if pd.notna(row["median_debt_usd"]):
        add_qa(
            idx,
            program,
            f"{context} What is the median student debt?",
            f"The reported median student debt is "
            f"${row['median_debt_usd']:,.0f} USD."
        )

    # Linked occupation
    if pd.notna(row["largest_linked_occupation"]):
        add_qa(
            idx,
            program,
            f"{context} Which occupation is most strongly linked to this program?",
            f"The largest linked occupation is "
            f"{row['largest_linked_occupation']}."
        )

    # Occupation wage
    if pd.notna(row["occupation_median_wage_2024_usd"]):
        add_qa(
            idx,
            program,
            f"{context} What is the median annual wage of its linked occupation?",
            f"The median annual wage of the linked occupation is "
            f"${row['occupation_median_wage_2024_usd']:,.0f} USD."
        )

    # Occupation growth
    if pd.notna(row["occupation_growth_pct_2024_34"]):
        add_qa(
            idx,
            program,
            f"{context} What is the projected occupation growth percentage from 2024 to 2034?",
            f"The projected occupation growth is "
            f"{row['occupation_growth_pct_2024_34']:.2f}%."
        )

    # AI tools
    if pd.notna(row["ai_tool_examples"]):
        add_qa(
            idx,
            program,
            f"{context} Which AI tools are associated with this occupation?",
            f"The associated AI tools include {row['ai_tool_examples']}."
        )


qa_df = pd.DataFrame(qa_data)

print("Question-answer dataset created!")
print("Total QA examples:", len(qa_df))

display(qa_df.head(10))
```

```python
from sklearn.model_selection import train_test_split

# Split unique original record IDs
record_ids = slm_df.index.to_numpy()

train_ids, temp_ids = train_test_split(
    record_ids,
    test_size=0.20,
    random_state=42
)

val_ids, test_ids = train_test_split(
    temp_ids,
    test_size=0.50,
    random_state=42
)

# Assign QA examples based on original record IDs
train_df = qa_df[qa_df["record_id"].isin(train_ids)].copy()
val_df = qa_df[qa_df["record_id"].isin(val_ids)].copy()
test_df = qa_df[qa_df["record_id"].isin(test_ids)].copy()

print("Training examples:", len(train_df))
print("Validation examples:", len(val_df))
print("Testing examples:", len(test_df))
```

```python
print("Training records:", train_df["record_id"].nunique())
print("Validation records:", val_df["record_id"].nunique())
print("Testing records:", test_df["record_id"].nunique())

# Check for overlapping records
train_records = set(train_df["record_id"])
val_records = set(val_df["record_id"])
test_records = set(test_df["record_id"])

print("\nTraining/Validation overlap:",
      len(train_records & val_records))

print("Training/Testing overlap:",
      len(train_records & test_records))

print("Validation/Testing overlap:",
      len(val_records & test_records))
```

```python
import torch
import pandas as pd
import numpy as np

from transformers import (
    AutoTokenizer,
    AutoModelForSeq2SeqLM,
    DataCollatorForSeq2Seq,
    Trainer,
    TrainingArguments
)

print("Hugging Face libraries imported successfully!")
print("PyTorch version:", torch.__version__)
print("Device:", "CUDA" if torch.cuda.is_available() else "CPU")
```

```python
model_name = "google/flan-t5-small"

print("Downloading tokenizer...")

tokenizer = AutoTokenizer.from_pretrained(model_name)

print("Downloading pretrained model...")

model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

print("Model downloaded successfully!")

print("Model parameters:",
      sum(p.numel() for p in model.parameters()))

print("Tokenizer vocabulary:",
      tokenizer.vocab_size)
```

```python
from sklearn.model_selection import GroupShuffleSplit

split1 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, temp_idx = next(
    split1.split(qa_df, groups=qa_df["program"])
)

train_df = qa_df.iloc[train_idx].copy()
temp_df = qa_df.iloc[temp_idx].copy()

split2 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.50,
    random_state=42
)

val_idx, test_idx = next(
    split2.split(temp_df, groups=temp_df["program"])
)

val_df = temp_df.iloc[val_idx].copy()
test_df = temp_df.iloc[test_idx].copy()

print("Training examples:", len(train_df))
print("Validation examples:", len(val_df))
print("Testing examples:", len(test_df))

print("\nProgram overlap:")
print("Train/Validation:",
      len(set(train_df.program) & set(val_df.program)))

print("Train/Test:",
      len(set(train_df.program) & set(test_df.program)))

print("Validation/Test:",
      len(set(val_df.program) & set(test_df.program)))
```

```python
def prepare_input(question):
    return "Answer the following question using the educational dataset: " + question

train_inputs = [
    prepare_input(q) for q in train_df["question"]
]

val_inputs = [
    prepare_input(q) for q in val_df["question"]
]

test_inputs = [
    prepare_input(q) for q in test_df["question"]
]

train_answers = train_df["answer"].tolist()
val_answers = val_df["answer"].tolist()
test_answers = test_df["answer"].tolist()

print("Training inputs prepared:", len(train_inputs))
print("Validation inputs prepared:", len(val_inputs))
print("Testing inputs prepared:", len(test_inputs))

print("\nExample input:")
print(train_inputs[0])

print("\nExpected answer:")
print(train_answers[0])
```

```python
MAX_INPUT_LENGTH = 128
MAX_TARGET_LENGTH = 96

def tokenize_examples(inputs, answers):

    model_inputs = tokenizer(
        inputs,
        max_length=MAX_INPUT_LENGTH,
        truncation=True
    )

    labels = tokenizer(
        text_target=answers,
        max_length=MAX_TARGET_LENGTH,
        truncation=True
    )

    model_inputs["labels"] = labels["input_ids"]

    return model_inputs


train_encodings = tokenize_examples(
    train_inputs,
    train_answers
)

val_encodings = tokenize_examples(
    val_inputs,
    val_answers
)

test_encodings = tokenize_examples(
    test_inputs,
    test_answers
)

print("Tokenization completed!")

print("Training tokens:", len(train_encodings["input_ids"]))
print("Validation tokens:", len(val_encodings["input_ids"]))
print("Testing tokens:", len(test_encodings["input_ids"]))
```

```python
from torch.utils.data import Dataset

class EducationalDataset(Dataset):

    def __init__(self, encodings):
        self.encodings = encodings

    def __len__(self):
        return len(self.encodings["input_ids"])

    def __getitem__(self, index):

        return {
            key: torch.tensor(value[index])
            for key, value in self.encodings.items()
        }


train_dataset = EducationalDataset(train_encodings)
val_dataset = EducationalDataset(val_encodings)
test_dataset = EducationalDataset(test_encodings)

print("PyTorch datasets created!")

print("Training:", len(train_dataset))
print("Validation:", len(val_dataset))
print("Testing:", len(test_dataset))
```

```python
data_collator = DataCollatorForSeq2Seq(
    tokenizer=tokenizer,
    model=model,
    padding=True
)

print("Data collator ready!")
```

```python
import inspect

training_parameters = inspect.signature(
    TrainingArguments.__init__
).parameters

eval_parameter = (
    "eval_strategy"
    if "eval_strategy" in training_parameters
    else "evaluation_strategy"
)

training_config = {
    "output_dir": "./educareer_checkpoints",
    "num_train_epochs": 2,
    "per_device_train_batch_size": 2,
    "per_device_eval_batch_size": 2,
    "gradient_accumulation_steps": 4,
    "learning_rate": 5e-5,
    "weight_decay": 0.01,
    "logging_steps": 25,
    "save_strategy": "epoch",
    "load_best_model_at_end": True,
    "metric_for_best_model": "eval_loss",
    "greater_is_better": False,
    "report_to": "none",
    "dataloader_num_workers": 0
}

training_config[eval_parameter] = "epoch"

training_args = TrainingArguments(**training_config)

print("Training configuration created!")
```

```python
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
    data_collator=data_collator
)

print("Trainer initialized successfully!")
```

```python
print("EduCareer AI training started!")

training_result = trainer.train()

print("\nTraining completed!")

print("Training loss:",
      training_result.training_loss)
```

```python
validation_results = trainer.evaluate(
    eval_dataset=val_dataset
)

print("VALIDATION RESULTS")
print("-------------------")

for key, value in validation_results.items():
    if isinstance(value, (int, float)):
        print(key, ":", round(value, 4))
```

```python
model.eval()

def generate_answer(question):

    prompt = prepare_input(question)

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=MAX_INPUT_LENGTH
    )

    inputs = {
        key: value.to(model.device)
        for key, value in inputs.items()
    }

    with torch.no_grad():

        output = model.generate(
            **inputs,
            max_new_tokens=96,
            num_beams=2,
            early_stopping=True
        )

    return tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )


for i in range(min(5, len(test_df))):

    question = test_df.iloc[i]["question"]
    expected = test_df.iloc[i]["answer"]

    prediction = generate_answer(question)

    print("QUESTION:", question)
    print("\nEXPECTED:", expected)
    print("\nPREDICTED:", prediction)
    print("\n" + "="*70)
```

```python
from sklearn.metrics import accuracy_score

predictions = []
actual_answers = []

for question, answer in zip(
    test_df["question"],
    test_df["answer"]
):

    predictions.append(generate_answer(question))
    actual_answers.append(answer)


def normalize(text):
    return " ".join(str(text).lower().strip().split())


exact_matches = [
    normalize(p) == normalize(a)
    for p, a in zip(predictions, actual_answers)
]

exact_match_score = np.mean(exact_matches)

print("Exact Match Score:",
      round(exact_match_score * 100, 2), "%")

print("Correct exact matches:",
      sum(exact_matches))

print("Total test examples:",
      len(actual_answers))
```

```python
import pandas as pd

train_df = pd.read_csv("train_data.csv")
validation_df = pd.read_csv("validation_data.csv")
test_df = pd.read_csv("test_data.csv")

print("Training records:", len(train_df))
print("Validation records:", len(validation_df))
print("Testing records:", len(test_df))

print("\nDataset columns:")
print(train_df.columns.tolist())

display(train_df.head())
```

```python
from transformers import AutoTokenizer
from datasets import Dataset, DatasetDict

BASE_MODEL = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(BASE_MODEL)

def prepare_dataset(df):
    df = df.dropna(subset=["question", "answer"]).copy()

    df["input_text"] = (
        "Answer the following EduCareer question: "
        + df["question"].astype(str)
    )

    df["target_text"] = df["answer"].astype(str)

    return Dataset.from_pandas(
        df[["input_text", "target_text"]],
        preserve_index=False
    )

raw_datasets = DatasetDict({
    "train": prepare_dataset(train_df),
    "validation": prepare_dataset(validation_df),
    "test": prepare_dataset(test_df)
})

def tokenize_function(example):
    inputs = tokenizer(
        example["input_text"],
        max_length=128,
        truncation=True
    )

    targets = tokenizer(
        text_target=example["target_text"],
        max_length=96,
        truncation=True
    )

    inputs["labels"] = targets["input_ids"]

    return inputs

tokenized_datasets = raw_datasets.map(
    tokenize_function,
    batched=True,
    remove_columns=raw_datasets["train"].column_names
)

print(tokenized_datasets)
print("Tokenization completed successfully!")
```

```python
from torch.utils.data import DataLoader
from transformers import DataCollatorForSeq2Seq

data_collator = DataCollatorForSeq2Seq(
    tokenizer=tokenizer,
    padding=True,
    return_tensors="pt"
)

train_loader = DataLoader(
    tokenized_datasets["train"],
    batch_size=2,
    shuffle=True,
    collate_fn=data_collator
)

validation_loader = DataLoader(
    tokenized_datasets["validation"],
    batch_size=2,
    shuffle=False,
    collate_fn=data_collator
)

test_loader = DataLoader(
    tokenized_datasets["test"],
    batch_size=2,
    shuffle=False,
    collate_fn=data_collator
)

batch = next(iter(train_loader))

print("Input IDs:", batch["input_ids"].shape)
print("Attention Mask:", batch["attention_mask"].shape)
print("Labels:", batch["labels"].shape)

print("Batch loading completed successfully!")
```

```python
import pandas as pd

train_df = pd.read_csv("train_data.csv")
validation_df = pd.read_csv("validation_data.csv")
test_df = pd.read_csv("test_data.csv")

print("Training records:", len(train_df))
print("Validation records:", len(validation_df))
print("Testing records:", len(test_df))

print("\nColumns:", train_df.columns.tolist())

display(train_df.head())
```

```python
from transformers import AutoTokenizer
from datasets import Dataset, DatasetDict

# Load FLAN-T5 tokenizer
BASE_MODEL = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(BASE_MODEL)

print("FLAN-T5 tokenizer loaded successfully!")
print("Vocabulary size:", tokenizer.vocab_size)
```

```python
def prepare_dataset(df):

    df = df.dropna(subset=["question", "answer"]).copy()

    df["input_text"] = (
        "Answer the following EduCareer question: "
        + df["question"].astype(str)
    )

    df["target_text"] = df["answer"].astype(str)

    return Dataset.from_pandas(
        df[["input_text", "target_text"]],
        preserve_index=False
    )


raw_datasets = DatasetDict({

    "train": prepare_dataset(train_df),

    "validation": prepare_dataset(validation_df),

    "test": prepare_dataset(test_df)

})

print(raw_datasets)
```

```python
def tokenize_function(example):

    inputs = tokenizer(
        example["input_text"],
        max_length=128,
        truncation=True
    )

    targets = tokenizer(
        text_target=example["target_text"],
        max_length=96,
        truncation=True
    )

    inputs["labels"] = targets["input_ids"]

    return inputs


tokenized_datasets = raw_datasets.map(
    tokenize_function,
    batched=True,
    remove_columns=raw_datasets["train"].column_names
)

print("Tokenization completed successfully!")

print(tokenized_datasets)
```

```python
print("Example input IDs:")
print(tokenized_datasets["train"][0]["input_ids"])

print("\nExample target IDs:")
print(tokenized_datasets["train"][0]["labels"])

print("\nDecoded question:")
print(
    tokenizer.decode(
        tokenized_datasets["train"][0]["input_ids"],
        skip_special_tokens=True
    )
)

print("\nDecoded answer:")
print(
    tokenizer.decode(
        tokenized_datasets["train"][0]["labels"],
        skip_special_tokens=True
    )
)
```

```python
from torch.utils.data import DataLoader
from transformers import DataCollatorForSeq2Seq

# Prepare data collator
data_collator = DataCollatorForSeq2Seq(
    tokenizer=tokenizer,
    padding=True,
    return_tensors="pt"
)

# Training data loader
train_loader = DataLoader(
    tokenized_datasets["train"],
    batch_size=2,
    shuffle=True,
    collate_fn=data_collator
)

# Validation data loader
validation_loader = DataLoader(
    tokenized_datasets["validation"],
    batch_size=2,
    shuffle=False,
    collate_fn=data_collator
)

# Testing data loader
test_loader = DataLoader(
    tokenized_datasets["test"],
    batch_size=2,
    shuffle=False,
    collate_fn=data_collator
)

# Get one training batch
batch = next(iter(train_loader))

print("Input IDs shape:", batch["input_ids"].shape)
print("Attention Mask shape:", batch["attention_mask"].shape)
print("Labels shape:", batch["labels"].shape)

print("\nBatch loading completed successfully!")
```

```python
from transformers import AutoConfig

# Load FLAN-T5 configuration
config = AutoConfig.from_pretrained(BASE_MODEL)

print("Model:", config.model_type)
print("Architecture:", config.architectures)

print("Encoder Layers:", config.num_layers)
print("Attention Heads:", config.num_heads)

print(
    "Relative Attention Buckets:",
    config.relative_attention_num_buckets
)

print("\nFLAN-T5 uses relative position bias.")
print("It does not use the Qwen-style RoPE mechanism.")
```
